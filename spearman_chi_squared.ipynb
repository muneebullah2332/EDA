```python
import numpy as np
import matplotlib.pyplot as plt

import pandas as pd
import seaborn as sns
from pylab import rcParams

import scipy
from scipy.stats import spearmanr
```

```python
%matplotlib inline
rcParams['figure.figsize'] = 14, 7
sns.set_style('whitegrid')
```

```python
df = sns.load_dataset('car_crashes')
df.head()
```

```python
sns.pairplot(df)
```

```python
df.columns
```

```python
x = df[['total', 'speeding', 'alcohol', 'ins_losses']]
sns.pairplot(x)
```

```python
total = df['total']
speeding = df['speeding']
alcohol = df['alcohol']
ins_losses = df['ins_losses']
```

```python
spearman_coefficient, p_value = spearmanr(total, speeding)
print(f"Spearman's rank correlation coefficient: {spearman_coefficient}")
```

```python
spearman_coefficient, p_value = spearmanr(total, alcohol)
print(f"Spearman's rank correlation coefficient: {spearman_coefficient}")
```

```python
spearman_coefficient, p_value = spearmanr(total, ins_losses)
print(f"Spearman's rank correlation coefficient: {spearman_coefficient}")
```

```python
table = pd.crosstab(total, speeding)
from scipy.stats import chi2_contingency
chi2, p, dof, expected = chi2_contingency(table.values)
print(f"Chi-squared statistic: {chi2}, p-value: {p}, degrees of freedom: {dof}")
```

```python
table = pd.crosstab(total, alcohol)
from scipy.stats import chi2_contingency
chi2, p, dof, expected = chi2_contingency(table.values)
print(f"Chi-squared statistic: {chi2}, p-value: {p}, degrees of freedom: {dof}")
```
